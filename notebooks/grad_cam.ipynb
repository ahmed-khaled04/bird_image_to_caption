# Grad-CAM: scratch versus transfer learning
Run this after the team has trained both models and supplied checkpoints saved by `train_run`.
Grad-CAM averages the gradients of a class logit over each final feature map, weights the maps,
and applies ReLU. Warm colours indicate stronger positive influence on that class score.
Maps are normalized separately per image: colours are not comparable absolute importance scores.
